# ⚓ BITACORA DE RESPUESTAS — Capitulo II

Construccion paso a paso de metricas cruzadas con `origen_proceso`.

```python
import sys, os, pandas as pd

if os.path.abspath("..") not in sys.path:
    sys.path.insert(0, os.path.abspath(".."))

from database.db import engine

query = """
SELECT id, experiencia_anios, origen_proceso
FROM ofertas
"""
df = pd.read_sql(query, engine)
print(f"Carga exitosa: {len(df)} registros.")
```

## II.1 origen_proceso x experiencia_anios

```python
# II.1.1 Promedio por Origen
print("Promedio de experiencia por origen:")
print(df.groupby("origen_proceso")["experiencia_anios"].mean().round(2))
```

```python
# II.1.2 Mediana por Origen
print("Mediana de experiencia por origen:")
print(df.groupby("origen_proceso")["experiencia_anios"].median())
```

```python
# II.1.3 Moda por Origen
print("Moda de experiencia por origen:")
print(df.groupby("origen_proceso")["experiencia_anios"].apply(lambda x: x.mode()[0] if not x.mode().empty else None))
```

```python
# II.1.4 Distribucion de Niveles por Origen
bins = [0, 1.9, 4.9, 100]
labels = ["Junior (0-2)", "Middle (2-5)", "Senior (5+)"]
df["nivel"] = pd.cut(df["experiencia_anios"], bins=bins, labels=labels)

for origen in df["origen_proceso"].unique():
    subset = df[df["origen_proceso"] == origen]
    print(f"--- {origen} ---")
    print(subset["nivel"].value_counts())
    print()
```

```python
# II.1.5 Tasa Entry Level por Origen
print("% de ofertas sin experiencia (0 anos) por origen:")
for origen in df["origen_proceso"].unique():
    subset = df[df["origen_proceso"] == origen]
    rate = (subset["experiencia_anios"] == 0).mean() * 100
    print(f"{origen}: {rate:.2f}%")
```
