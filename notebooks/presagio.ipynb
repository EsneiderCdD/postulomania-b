# 🔮 PRESAGIO: Hoja de Ruta Técnica para la Plataforma
Este documento establece las directrices de arquitectura y el stack tecnológico para la evolución del proyecto Postulomaniaco hacia una plataforma web funcional.

## 1. Arquitectura de Datos para el Frontend
La estrategia de comunicación entre la base de datos y la interfaz se basa en una **Arquitectura de Agregación y Desacople** para optimizar el rendimiento y la experiencia de usuario.

### Endpoints de Especialidad (Evitar el "Big GET")
El backend debe procesar la lógica de negocio y entregar datos ya agregados. Se definen endpoints específicos para cada necesidad visual:
- `GET /api/v1/stats/tech-stack`: Retorna un objeto JSON con el ranking de tecnologías.
- `GET /api/v1/stats/experience-levels`: Retorna la distribución porcentual de niveles de cargo.
- `GET /api/v1/offers/search`: Retorna resultados paginados de ofertas detalladas.

### Data Contracts (JSON sobre Multimedia)
El backend debe servir exclusivamente datos en formato JSON (materia prima). La renderización de gráficos (SVG/Canvas) debe delegarse al frontend mediante librerías especializadas (Recharts, Chart.js). Esto garantiza:
- Interactividad (tooltips, filtros dinámicos).
- Menor consumo de ancho de banda.
- Capacidad de respuesta (Responsive Design).

### Estrategia de Pre-calculado
Dado que la actualización de datos depende de la ejecución del scraper y no de interacciones de usuario en tiempo real, se implementará una capa de persistencia para estadísticas:
- **Materialized Views** o **Tablas de Cache**: Al finalizar el proceso de scraping, una tarea automatizada debe regenerar la "foto del día".
- La API leerá directamente estos valores pre-calculados, reduciendo la latencia de respuesta a milisegundos.

## 2. Ecosistema Tecnológico Recomendado
Para mantener la coherencia con la base de datos PostgreSQL y el motor de análisis en Python, se establece el siguiente stack profesional:

### Backend: FastAPI (Python)
- **Justificación**: Alto rendimiento (asíncrono), validación de tipos nativa con Pydantic y generación automática de documentación (OpenAPI/Swagger).
- **Ventaja**: Permite reutilizar toda la lógica de `mining_stats/metrics.py` y los modelos de SQLAlchemy sin cambiar de lenguaje de programación.

### Frontend: Next.js + TypeScript + Tailwind CSS
- **Next.js**: Estándar industrial para aplicaciones React escalables con Server Side Rendering (SSR).
- **TypeScript**: Garantiza la integridad de los contratos de datos entre el API de Python y la interfaz.
- **Tailwind CSS**: Framework de utilidades para un diseño visualmente premium y consistente.

### Visualización y UI
- **Componentes**: **Shadcn UI**. Librería de componentes accesibles y personalizables con estética moderna.
- **Gráficos**: **Tremor.so**. Librería de componentes React optimizada para dashboards de datos y analíticas.
