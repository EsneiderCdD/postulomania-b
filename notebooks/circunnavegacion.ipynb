# Catálogo de Análisis Estadístico - Master Analytics
Este notebook sirve como hoja de ruta y checklist para los análisis del proyecto.

## 1. Origen del Proceso (`origen_proceso`)
Identifica el flujo o proceso de recolección de la oferta laboral.

- [x] **[Frecuencia]** ¿Cuál es el volumen total de ofertas por cada origen de proceso?
- [x] **[Distribución]** ¿Qué porcentaje del total representa cada origen?
- [x] **[Moda]** ¿Cuál es el origen de proceso con mayor cantidad de registros recolectados?
- [x] **[Calidad]** ¿Existen orígenes que presenten una mayor proporción de valores nulos?

## 2. Empresa (`empresa`)
Nombre de la organización o agencia que publica la vacante.

- [x] **[Frecuencia]** ¿Cuáles son las 10 empresas con mayor cantidad de ofertas publicadas?
- [x] **[Integridad]** ¿Cuál es el porcentaje de ofertas con nombre de empresa no especificado?
- [x] **[Segmentación]** ¿Qué empresas publican ofertas exclusivamente con requerimiento de inglés?
- [x] **[Unicidad]** ¿Cuántas empresas distintas existen en total?
- [x] **[Larga Cola]** ¿Qué porcentaje de empresas publican una sola oferta?


## 3. Tech Stack (`tech_stack`)
Lista de tecnologías, lenguajes y frameworks extraídos.

- [x] **[Popularidad]** ¿Cuáles son las 15 tecnologías más demandadas en el mercado?
- [x] **[Combinación]** ¿Qué pares de tecnologías co-ocurren con mayor frecuencia?
- [x] **[Promedio]** ¿Cuál es la cantidad promedio de tecnologías requeridas por oferta?
- [x] **[Rareza]** ¿Qué tecnologías aparecen en menos del 1% de las ofertas totales?
- [x] **[Tendencia]** ¿Cómo varía el stack tecnológico según el origen del proceso?
- [x] **[Unicidad]** ¿Cuántas tecnologías distintas hemos identificado en total?
- [x] **[Densidad]** ¿Cuál es el rango (mínimo y máximo) de tecnologías por oferta?


## 4. Experiencia en Años (`experiencia_anios`)
Años mínimos de experiencia profesional requeridos.

- [x] **[Promedio]** ¿Cuál es la media aritmética de años de experiencia requeridos?
- [x] **[Mediana]** ¿Cuál es el valor central de años de experiencia para evitar sesgos?
- [x] **[Distribución]** ¿Cómo se distribuyen las ofertas según niveles (Junior, Middle, Senior, Lead)?
- [x] **[Correlación]** ¿Existe relación entre el número de tecnologías y los años de experiencia?
- [x] **[Extremos]** ¿Cuál es el valor máximo solicitado y para qué tecnologías?
- [x] **[Entry Level]** ¿Qué porcentaje de ofertas no piden experiencia previa (0 años)?
- [x] **[Volatilidad]** ¿Qué tan dispersos están los años de experiencia (Desviación Estándar)?
- [x] **[Moda]** ¿Cuál es el valor exacto de años más repetido en el mercado?


## 5. Compatibilidad (`compatibilidad`)
Puntaje de ajuste de la oferta al perfil del usuario.

- [x] **[Promedio]** ¿Cuál es la compatibilidad media de todas las ofertas analizadas?
- [x] **[Top]** ¿Cuántas ofertas superan el umbral de compatibilidad del 70%?
- [x] **[Desempeño]** ¿Qué origen de proceso entrega las ofertas con mayor compatibilidad?
- [x] **[Dispersión]** ¿Cuál es la desviación estándar de la compatibilidad en el dataset?
- [x] **[Mediana]** ¿Cuál es el valor central de compatibilidad para evitar el sesgo de promedios?
- [x] **[Descarte]** ¿Qué porcentaje de ofertas son directamente inútiles (compatibilidad < 40%)?
- [x] **[Rango Absoluto]** ¿Cuáles son los valores mínimos y máximos reales arrojados por el sistema?


## 6. Requerimiento de Inglés (`requiere_ingles`)
Indicador booleano de obligatoriedad del idioma inglés.

- [x] **[Proporción]** ¿Qué porcentaje de las ofertas totales exigen inglés?
- [x] **[Impacto]** ¿Cómo afecta el requerimiento de inglés a los años de experiencia?
- [x] **[Relación]** ¿Qué tecnologías están más ligadas a ofertas bilingües?
- [x] **[Concentración]** ¿Qué origen de proceso tiene el mayor porcentaje de ofertas bilingües?
- [x] **[Carga Técnica]** ¿Las ofertas bilingües exigen dominar mayor cantidad de tecnologías?
- [x] **[Match Profile]** ¿El perfil actual hace mejor match con ofertas en inglés o locales?


## 7. Fecha de Publicación Estimada (`fecha_publicacion_estimada`)
Marca temporal de publicación de la oferta.

- [x] **[Volumen Temporal]** ¿Cuántas ofertas se publican por día de la semana?
- [x] **[Recencia]** ¿Qué cantidad de ofertas han sido publicadas en las últimas 24 y 48 horas?
- [x] **[Frecuencia]** ¿En qué franjas horarias se detecta mayor actividad de publicación?
- [x] **[Antigüedad Máxima]** ¿Cuántos días de antigüedad tiene la oferta más vieja (Frescura)?
- [x] **[Día Pico]** ¿Cuál es matemática y estadísticamente el mejor día para buscar trabajo?
- [x] **[Fuga de Fin de Semana]** ¿Qué porcentaje del total se publica sábados y domingos?


## 8. Título de la Oferta (`titulo`)
Denominación oficial del cargo publicado. (Campo especial que es mejor trabajar en v.2)

- [ ] **[Nube de Palabras]** ¿Cuáles son los términos más frecuentes en los nombres de los cargos?

Identificador único de la vacante.

- [x] **[Unicidad]** ¿Cuántos registros únicos existen en el maestro?
- [x] **[Duplicidad]** ¿Se han detectado IDs repetidos provenientes de distintos orígenes? (Incluye Tasa %)
- [x] **[Volumen]** ¿Cuál es el recuento total de registros para validación de integridad?
