# Auditoría de Seguridad: Error Leakage en API

**Alcance:** Global — todas las rutas en `api/routes/stats.py` y `api/routes/ofertas.py`.

**Estado:** Pendiente de corrección.

## ¿Por qué `return {"error": str(e)}` es un problema?

Actualmente si algo falla en cualquier endpoint, el cliente recibe el mensaje crudo del error de Python. Ejemplos de lo que podría filtrarse:

- `"Table 'ofertas' doesn't exist"` → expone estructura de BD
- `"could not connect to server: Connection refused"` → expone infraestructura
- `"invalid input syntax for type integer"` → expone esquema de columnas
- `"name 'x' is not defined"` → expone nombres de variables internas

En producción eso le da información gratis a un atacante sobre tu stack. La práctica segura es loggear el error internamente y devolver un mensaje genérico al cliente:

```json
{"error": "Error interno del servidor"}
```

En el caso actual (localhost, desarrollo), no hay riesgo real. Pero es buena práctica corregirlo ahora que se está ordenando el código.
