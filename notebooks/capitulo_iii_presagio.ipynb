## Desafios encontrados1. **Volumen de datos**: 58 ofertas son suficientes para la mayoria de cruces, pero no para patrones temporales finos (hora del dia) ni para rankings de empresas con una sola oferta.2. **Datos nulos**: Empresas sin nombre (~15%) limitan los cruces con `empresa`. Se filtran con `dropna()` en lugar de imputar.3. **Compatibilidad cero**: El perfil actual (`user_profile.json`) produce score 0 para ofertas con ingles (nivel de ingles del perfil = 0.3). Esto distorsiona `tech-compatibility` y `empresa-compatibility`.4. **Outliers**: La oferta con 31 anos de experiencia distorsiona promedios. Se usa mediana como complemento.
