# 📬 CONTACTAR — Planificación y RetosDocumento de conclusiones tras la conversación sobre el nuevo departamento **CONVERSIÓN**.Aquí se registran únicamente los retos, planteamientos e hitos identificados.No contiene soluciones ni implementaciones. Es una base para revisar, complementar y luego resolver.

---## 1. SCRAPER — Nuevos retosEl scraper actual (`dds_antioquia`) funciona, pero para alimentar CONVERSIÓN se necesitan ajustes:### 1.1 Ampliación de términos de búsqueda- Agregar palabras clave adicionales: **Trainee, Practicante, Pasantía, Junior**.- Evaluar a profundidad los términos ya usados y las ofertas extraídas para mejorar la precisión.- Evitar que el scraper capture ofertas no deseadas cuando Computrabajo muestra mensajes como:  - "No hay más ofertas de desarrollador de software en..."  - "Mira estas oportunidades de trabajo remoto en todo el país"  - "Ya viste todas las ofertas. Estas opciones también podrían interesarte"- El scraper es visible siempre; hay que enseñarle a reconocer estos casos de uso para no filtrarlos erróneamente.### 1.2 Frecuencia de ejecución- Ajustar la localización a revisión diaria o cada 3 días para capturar ofertas no vistas.- Mantener revisión de 1 sola página (la más reciente).- Postularse de primeras es clave: la información debe ser rápida y fresca.- Adaptar la frecuencia sin ser bloqueados por la plataforma.- Escalar a más revisiones sin que sea un problema, volviéndolo rutinario.- Ajustar las horas de ejecución usando las métricas ya existentes (días, horas pico de publicación).### 1.3 Coherencia con la fase REMOTO- Fase 1 actual: **Antioquia** (presencial y/o remoto dentro del departamento).- Fase 2 planeada: **Remoto sin restricción de Antioquia**.- El scraper debe mantenerse coherente entre fases: ¿aumentar scrapers separados o cambiar el flujo de 1 solo?- Reto: poblar el modelo sin romperlo al agregar REMOTO.- Incluir datos personales y búsquedas de manera global cuando corresponda.

---## 2. DATOS — Nuevos retosSe necesita registrar nueva información sobre las empresas y las ofertas, más allá de lo que ya existe.### 2.1 Sitio web y coordenadas (verificabilidad y seguridad)- Registrar si la empresa **tiene sitio web**: indicador de seriedad / verificabilidad.- Registrar si la empresa **tiene coordenadas** (lat/lng): indicador de presencia física real.- Estos datos funcionan como patrones de **verificabilidad y seguridad** a nivel de UI / puntuación.- Guardar el dato con una perspectiva hacia puntuar, asignar una nota o score de confianza.### 2.2 Agencia vs. Empresa directa- Clasificar si la oferta proviene de una **agencia de empleo/outsourcing** o de una **empresa directa**.- Implican dos flujos de conversión diferentes:  - **Agencia**: lo normal es registrarse en su portal y hacer un proceso interno. Hay que analizar la propuesta y su calidad para decidir si vale la pena.  - **Empresa directa**: sujeta a los análisis que se encuentren (contacto, respuesta, etc.).- Reto: cada flujo requiere soluciones distintas; se debe documentar y pulir en el camino.### 2.3 Historial de interacciones- Registrar si ya he hecho postulaciones a esta empresa antes.- ¿Cuándo? ¿Cuántas veces?- El historial permite medir la reputación de la empresa (¿responde? ¿nunca responde?).- Métricas de frecuencia de publicación por empresa ya existen — aprovecharlas.### 2.4 Reutilización de datos existentes- Empresa, oferta, tecnologías, compatibilidad, coordenadas, fechas: ya existen.- Determinar qué se reutiliza tal cual y qué requiere nuevas tablas o columnas.- Las métricas de frecuencia, fechas y horas ya están — suficientes por ahora.

---## 3. CONTACTO — RetosEl objetivo no es un contacto genérico (empresa, reclutador, origen, email como campos sueltos).La meta es encontrar un **recurso humano** por medios digitales, no solo dejar la interaccióna medios digitales como formularios web o postulaciones por la plataforma.### 3.1 Contacto personal- Se busca un **contacto personal**, no un contacto abstracto o genérico.- El medio es digital (email, LinkedIn, website), pero el destinatario debe ser una persona.- Reto: ¿cómo llegar al recurso humano real detrás de la oferta?- No es lo mismo un formulario de postulación web que un email directo a un reclutador.### 3.2 Intermediarios- Si hay un intermediario (agencia o persona), este tendrá sus propios datos.- Cada flujo (agencia, empresa directa, freelance) debe analizarse por separado.- La documentación debe describir la tarea correctamente, sin simular datos ni asumir flujos.- Reto: mantener el mayor control posible y veracidad de los datos en cada tipo de contacto.### 3.3 Origen del contacto- El contacto puede provenir de:  - Scraping del sitio web de la empresa (página /contacto, /trabaja-con-nosotros).  - Búsqueda externa (LinkedIn, directorios).  - La propia oferta (poco frecuente en Computrabajo).- Reto: automatizar la búsqueda del contacto sin depender de formatos estándar (cada sitio es distinto).- Reto: verificar si el contacto es válido (¿el email rebota?).

---## 4. ARQUITECTURA — Lo que ya existe vs. lo que faltaAntes de construir modelos o rutas, discernir qué podemos reutilizar y qué necesita ser nuevo.### 4.1 Lo que ya existe (reutilizable)- `empresas`: nombre, lat, lng, direccion, municipio, departamento, website (campo existe pero vacío).- `ofertas`: id_oferta, titulo, enlace, descripcion, municipio, departamento, fechas, experiencia_anios, requiere_ingles, keyword, origen_proceso, empresa_id.- `tecnologias` + `ofertas_tecnologias`: stack tecnológico de cada oferta.- `compatibilidades`: score de afinidad técnica (0-1).- `postulaciones`: registro de postulaciones hechas en Computrabajo (estados de la plataforma).- Métricas: 20+ endpoints de stats ya implementados.- Mapa: endpoints `/mapa/ofertas` y `/mapa/empresas` con lat/lng.### 4.2 Lo nuevo que se necesita (a evaluar)- Clasificación de empresa: agencia vs. directa.- Tabla o campo para el **historial de interacciones** (contactos previos, respuestas).- Tabla o campo para el **contacto personal** encontrado (nombre, email, origen, fecha).- Tabla de **seguimiento / bitácora de eventos** de conversión (ver sección 5).- Campo `modalidad` en ofertas (presencial, remoto, híbrido) — hoy no se persiste.- Lógica de geocodificación automática (Nominatim) al insertar empresa nueva.- Scraper de sitio web de empresa para extraer email de contacto.- Fórmula de **Invertibilidad** como complemento a Compatibilidad (ver sección 6).

---## 5. FLUJO DE CONVERSIÓN — Ciclo de vidaA diferencia de `postulaciones` (que registra lo que ocurre **dentro** de Computrabajo),CONVERSIÓN registra lo que ocurre **por fuera**: investigo, contacto, envío, mido respuesta.Cada paso es una tarea segmentada. No son estados fijos, son **eventos fechados** que avanzan en el tiempo.Cada uno requiere un abordaje distinto.| Paso | Quién actúa | Qué pasó | Retos asociados ||------|------------|----------|-----------------|| **Descubierta** | El scraper | La oferta entró al sistema, es relevante y cercana | Términos de búsqueda, frecuencia, evitar ofertas no deseadas, filtro por departamento || **Investigada** | Yo (o script) | Busqué el sitio web, encontré o no contacto, clasifiqué la empresa (agencia/directa), verifiqué coordenadas | Scraping de sitio web, clasificación agencia/empresa, geocodificación, extracción de email || **Contactada** | Yo | Envié CV + propuesta personalizada por email o LinkedIn | Encontrar el contacto personal (recurso humano), redactar propuesta, evitar formularios genéricos || **En espera** | El tiempo | Han pasado X días sin respuesta | Definir umbral de espera, registrar intentos de seguimiento, decidir si re-contactar || **Respondida** | La empresa | Me contestaron (positivo, negativo o acuse de recibo) | Clasificar el tipo de respuesta, registrar fecha y contenido, medir tiempo de respuesta || **No respondida** | La empresa (por omisión) | Pasó el umbral de tiempo sin respuesta | Marcar como contacto fallido, ¿reintentar por otra vía?, actualizar reputación de la empresa || **Entrevista** | Ambos | Se concretó y se realizó la entrevista | Agendar, preparar, registrar resultado, ¿hubo feedback posterior? || **Cerrada** | Yo | Proceso terminado (con o sin éxito, o descartado por mí) | Registrar motivo de cierre, lecciones aprendidas, actualizar historial |**Nota:** Cada paso es un segmento independiente. Se aborda, se analiza y se resuelve por separado.La tabla sirve como guía del recorrido completo, no como una máquina de estados.

---## 6. INVERTIBILIDAD — Concepto complementario a CompatibilidadAsí como **Compatibilidad** mide qué tan apto soy yo para la oferta (nivel técnico),**Invertibilidad** mediría qué tan rentable es invertir mi esfuerzo en **contactar** esa oferta.### Factores propuestos (a refinar):- **Compatibilidad** (ya existe, 0-1): qué tan calificado estoy.- **Cercanía** (requiere geocodificación, 0-1): distancia a la empresa (1 si <5km, decae).- **Contactabilidad** (nuevo, 0-1): ¿tiene website? ¿encontré email de contacto?- **Reputación** (nuevo, 0-1): ¿respondió antes esta empresa? ¿es directa o agencia?Una oferta con compatibilidad 0.9 pero sin email y a 300km tendría baja invertibilidad.Una con compatibilidad 0.5 pero cercana, con contacto y que ya respondió antes tendría alta invertibilidad.**Reto:** Diseñar una fórmula mínima, básica, escalable, que sirva como instrumento de priorización,no como verdad absoluta.

---## 7. MÉTRICAS PROPIAS DE CONVERSIÓNA diferencia de las métricas actuales (que describen el mercado), estas describirían **mi interacción con el mercado**:- **Tasa de contacto encontrado**: ¿en cuántas ofertas investigadas hallé un email o persona de contacto?- **Tasa de respuesta**: de las contactadas, ¿cuántas respondieron (positiva o negativamente)?- **Tasa por tipo de empresa**: ¿agencias responden más o menos que empresas directas?- **Tiempo medio de respuesta** (días entre contacto y respuesta).- **Empresas respondedoras**: las que respondieron al menos una vez → alto valor, así no contraten.- **Empresas fantasma**: publican mucho, nunca responden → marcar para descartar en futuras iteraciones.- **Brecha real del mercado local**: ¿qué techs piden las que SÍ responden? ¿qué experiencia? → retroalimenta `user_profile.json`.**Nota:** La mayoría de estas métricas requieren datos que aún no existen (contacto, respuesta, etc.).Son aspiracionales y se irán construyendo conforme avance el flujo.
